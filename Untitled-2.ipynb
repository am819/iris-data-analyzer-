import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

# Convert to DataFrame
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Adding target column to the DataFrame
iris_df['species'] = iris.target

# Map target numbers to species names
iris_df['species'] = iris_df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

```python
plt.figure(figsize=(12, 8))

for i, bins in enumerate([5, 10, 15, 20]):
    plt.subplot(2, 2, i + 1)
    sns.histplot(iris_df['petal length (cm)'], bins=bins, kde=True)
    plt.title(f'Petal Length Histogram (bins={bins})')

plt.tight_layout()
plt.show()

# Display the Iris dataset DataFrame
print(iris_df)
```
